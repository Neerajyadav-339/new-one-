```python
s1 = {10,20,30,40}
s2 = {30,20,60,50}
print(s1 | s2) # this can fine the union 
print(s1.union(s2))
```

```python
print(s1 & s2)
print(s1.intersection(s2))
```

```python
# set diffenrence 
print(s1-s2)
# the number in the s1 and not in s2 
```

- Dictionary 

```python
d = {'id':28, 'name': 'Vihwas', 'dept': 'Developer', 'salary':20000}
print(type(d))
```

```python
print(d['salary'])
```

```python
d['salary'] = 40000 # keys are aslway mutable and values are mutable 
print(d)
```

*dictionary functions*

```python
print(d.keys())
print(d.items())
print(d.values())
```

```python
# fromkeys
key = d.keys()
new = dict.fromkeys(key) # this will take all keys form the dictionary and make another one 
print(new)
```

```python
# Remove key value pair 
d['dept'] = 'developer'
print(d)
d.pop('dept')
```

```python
print(d)
```

- Traverse dict

```python
for i in d:
    print(i," : ", d[i])
```

## Guessing the number

```python
import random

number = random.randint(1,10)
attempt = 0
while True:
    you = int(input("Guess the number:- "))
    attempt+=1
    if you == number:
        print(f"Yey you guessed the {number} in {attempt} !")
        break
    elif you > number:
        print("You Are too High !")
    elif you < number:
        print("You are too low !")
    
```
